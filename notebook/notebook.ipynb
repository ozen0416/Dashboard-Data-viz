```python
import pandas as pd

ANNEE_MIN, ANNEE_MAX = 1982, 2024   

df_catnat["code_commune"] = (df_catnat["code_geographique"].astype(str)
                             .str.replace(r"\.0$", "", regex=True).str.zfill(5))

for c in ["date_debut", "date_fin", "date_arrete"]:
    df_catnat[c] = pd.to_datetime(df_catnat[c], errors="coerce")
df_catnat["annee"] = df_catnat["date_debut"].dt.year

# Fenêtre d'analyse (garde toute la France pour le benchmark)
cn = df_catnat[df_catnat["annee"].between(ANNEE_MIN, ANNEE_MAX)].copy()
cn["annee"] = cn["annee"].astype(int)

# Vérifie le libellé exact, puis filtre
print(cn.loc[cn["libelle_epci"].str.contains("Bordeaux", case=False, na=False), "libelle_epci"].unique())
bm = cn[cn["libelle_epci"].str.contains("Bordeaux Métropole", case=False, na=False)].copy()
print(bm["code_commune"].nunique())   # attendu : 28
```

```python
MAP = {"Sécheresse": "Sécheresse / argiles",
       "Retrait-gonflement des argiles": "Sécheresse / argiles",
       "Inondations": "Inondations",
       "Cyclones / Tempêtes": "Tempêtes",
       "Mouvements de terrain": "Mouvements de terrain"}
for d in (cn, bm):
    d["groupe"] = d["type_catastrophe"].map(MAP).fillna("Autres")

# Vérification : les deux libellés "sécheresse" se succèdent-ils dans le temps ?
print(pd.crosstab(bm["annee"], bm["type_catastrophe"])[["Sécheresse", "Retrait-gonflement des argiles"]]
      if {"Sécheresse", "Retrait-gonflement des argiles"} <= set(bm["type_catastrophe"]) else "un des libellés est absent")

# Graphique 3 : épisodes distincts de la métropole par année et par type
episodes = bm.drop_duplicates(["groupe", "date_debut", "date_fin"])
ep_annee = episodes.groupby(["annee", "groupe"]).size().unstack(fill_value=0)

# Graphique 5 : heatmap commune × type (nombre d'arrêtés)
heat = bm.groupby(["libelle_geographique", "groupe"]).size().unstack(fill_value=0)
heat["total"] = heat.sum(axis=1)
heat = heat.sort_values("total", ascending=False)

# Graphique 4 : arrêtés moyens par commune (benchmark)
g = cn.groupby("libelle_epci").agg(nb=("code_commune", "size"),
                                   communes=("code_commune", "nunique"))
g["arretes_par_commune"] = g["nb"] / g["communes"]
metropoles = g[g.index.str.contains("métropole", case=False)]["arretes_par_commune"]

gir = cn[cn["departement"].astype(str).str.zfill(2) == "33"]
ref = pd.Series({"Gironde": len(gir) / gir["code_commune"].nunique(),
                 "France": len(cn) / cn["code_commune"].nunique()})
benchmark = pd.concat([metropoles, ref]).sort_values(ascending=False)
```

```python
AGG = {"TX": "mean", "NBJTX30": "sum", "NBJTX35": "sum", "NBJTNS20": "sum",
       "RR": "sum", "ETP": "sum", "NBJRR30": "sum", "NBJRR50": "sum"}
cols = [c for c in AGG if c in df_meteo.columns]
for c in cols:
    df_meteo[c] = pd.to_numeric(df_meteo[c], errors="coerce")
df_meteo["annee"] = df_meteo["AAAAMM"].astype(int) // 100

def annuel(col):
    g = df_meteo.groupby(["NUM_POSTE", "annee"])[col].agg([AGG[col], "count"]).reset_index()
    return g[g["count"] == 12]          # années complètes uniquement

parts = {c: annuel(c) for c in cols}

# Stations longues : au moins 40 années complètes (évite les artefacts de stations qui ouvrent/ferment)
ref_col = "NBJTX30" if "NBJTX30" in parts else cols[0]
nb_ans = parts[ref_col].groupby("NUM_POSTE").size()
stations_ok = nb_ans[nb_ans >= 40].index
print(len(stations_ok), "stations retenues")

meteo_annee = pd.concat(
    {c: p[p["NUM_POSTE"].isin(stations_ok)].groupby("annee")[AGG[c]].mean()
     for c, p in parts.items()}, axis=1)

if "ETP" in meteo_annee and meteo_annee["ETP"].notna().any():
    meteo_annee["bilan_hydrique"] = meteo_annee["RR"] - meteo_annee["ETP"]
for c in ["NBJTX30", "TX"]:
    if c in meteo_annee:
        meteo_annee[c + "_mm10"] = meteo_annee[c].rolling(10, min_periods=5).mean()

# Lien climat ↔ catastrophes (pour le nuage de points)
lien = meteo_annee.join(ep_annee, how="inner")
```
